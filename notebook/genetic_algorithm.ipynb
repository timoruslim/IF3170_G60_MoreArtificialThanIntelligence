```python
draw_state(init_state)
```

```python
obj = MaxValueObjective()
solver_ga = GeneticAlgorithm(obj)
result = solver_ga.solve(init_state)
print(result)
plt.figure()
plt.plot(result.metadata["max_history"], label="Maximum")
plt.plot(result.metadata["average_history"], label="Average")
plt.title(f"Objective History (population=6, iterations=100)")
plt.xlabel("Iterations")
plt.ylabel("Objective")
plt.legend()
plt.show()
```

```python
draw_state(result.final_state)
```
