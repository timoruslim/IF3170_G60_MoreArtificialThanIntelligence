```python
# Creating The Problem Set Initial State for Test 1
# Inputting Trucks with Truck(id, width, length, height, max_capacity)
trucks = [
   Truck("T1", 10, 20, 5, 200),
   Truck("T2", 8, 12, 4, 150),
   Truck("T3", 6, 10, 3, 100),
]

# Inputting Packages with Package(id, width, length, height, value, weight, is_fragile, eta) 
pkgs = [
   # Low value packages with light weight and small dimension
   Package("S1", 1, 1, 1, 50, 2, False, 1),
   Package("S2", 1, 2, 1, 45, 3, False, 3),
   Package("S3", 2, 1, 1, 40, 2, True, 2),
   Package("S4", 1, 1, 2, 35, 1, False, 4),
   Package("S5", 2, 2, 1, 55, 4, False, 1),

   # Moderate value packages with moderate-light weight and moderate dimension
   Package("M1", 3, 4, 2, 80, 10, False, 5),
   Package("M2", 2, 3, 2, 70, 8, True, 2),
   Package("M3", 4, 3, 1, 65, 12, False, 3),
   Package("M4", 3, 3, 3, 90, 15, False, 6),
   Package("M5", 2, 4, 2, 75, 9, False, 1),
   Package("M6", 3, 2, 2, 60, 11, True, 4),
   Package("M7", 4, 2, 3, 85, 14, False, 2),

   # High value packages with moderate weight and large dimension
   Package("L1", 5, 6, 3, 120, 25, False, 7),
   Package("L2", 4, 5, 2, 100, 20, False, 3),
   Package("L3", 6, 4, 2, 110, 22, True, 5),
   Package("L4", 5, 5, 3, 130, 30, False, 2),
   Package("L5", 4, 4, 4, 95, 18, False, 4),

   # High value packages with heavy weight and small dimension
   Package("H1", 3, 3, 2, 140, 35, False, 1),
   Package("H2", 2, 2, 2, 105, 28, False, 6),

   # Low value packages with heavy weight and large dimension
   Package("B1", 6, 8, 3, 30, 40, False, 7),
   Package("B2", 5, 7, 4, 25, 35, False, 5),

   # High value packages with light weight and small dimension
   Package("F1", 2, 3, 1, 115, 6, True, 1),
   Package("F2", 3, 2, 2, 100, 7, True, 1),
]

# Creating an initial state of packages and trucks with random but feasible orientation
initial_state = State(pkgs, trucks, init="random_feasible")

# Visualizing the packages position relative to the trucks
draw_state(initial_state)
```

```python
# Algorithm 1: Hill Climbing
from src.engine import *
from src.solvers.hill_climbing import SteepestAscent, Stochastic, SidewaysMove, RandomRestart
import matplotlib.pyplot as plt

# Set up requirements for hill climbing
obj = MaxValueObjective()
```

```python
# Algorithm 1.1: Steepest Ascent Hill Climbing
solver_steepest_ascent = []
results_steepest_ascent = []
for repetition in range(3):
    start_state = initial_state
    solver_steepest_ascent.append(SteepestAscent(obj, max_iterations = 100))
    results_steepest_ascent.append(solver_steepest_ascent[-1].solve(start_state))
    print(f"Results of repetition {repetition+1}:")
    print(results_steepest_ascent[repetition]) # Output: Runtime, Total Iterations, Initial Objective, Final Objective
    
    plt.plot(results_steepest_ascent[repetition].objective_history) # Output: Plot of Objective Value over Iterations
    
    draw_state(results_steepest_ascent[repetition].final_state) # Output: Visualization of Final State
    print("=" * 50)

plt.title('Objective History')
plt.xlabel('Iterations')
plt.ylabel('Objective')
plt.show()
```

```python
# Algorithm 1.2: Stochastic Hill Climbing
solver_stochastic = []
results_stochastic = []
for repetition in range(3):
    start_state = initial_state
    solver_stochastic.append(Stochastic(obj, max_random_successor_iterations=500, max_iterations = 100))
    results_stochastic.append(solver_stochastic[-1].solve(start_state))
    print(f"Results of repetition {repetition+1}:")
    print(results_stochastic[repetition]) # Output: Runtime, Total Iterations, Initial Objective, Final Objective, Random Successor Iterations
    print("Random Successor Iterations:", results_stochastic[repetition].metadata["random_successor_iterations"]) # Output: Number of Random Successor Iterations
    
    plt.plot(results_stochastic[repetition].objective_history) # Output: Plot of Objective Value over Iterations
    
    draw_state(results_stochastic[repetition].final_state) # Output: Visualization of Final State
    print("=" * 50)

plt.title('Objective History')
plt.xlabel('Iterations')
plt.ylabel('Objective')
plt.show()
```

```python
# Algorithm 1.2: Stochastic Hill Climbing, Trying More Iterations
solver_stochastic = []
results_stochastic = []
for repetition in range(3):
    start_state = initial_state
    solver_stochastic.append(Stochastic(obj, max_random_successor_iterations=500, max_iterations = 10000))
    results_stochastic.append(solver_stochastic[-1].solve(start_state))
    print(f"Results of repetition {repetition+1}:")
    print(results_stochastic[repetition]) # Output: Runtime, Total Iterations, Initial Objective, Final Objective, Random Successor Iterations
    print("Random Successor Iterations:", results_stochastic[repetition].metadata["random_successor_iterations"]) # Output: Number of Random Successor Iterations
    
    plt.plot(results_stochastic[repetition].objective_history) # Output: Plot of Objective Value over Iterations
    
    draw_state(results_stochastic[repetition].final_state) # Output: Visualization of Final State
    print("=" * 50)

plt.title('Objective History')
plt.xlabel('Iterations')
plt.ylabel('Objective')
plt.show()
```

```python
# Algorithm 1.3: Sideways Move Hill Climbing
solver_sideways = []
results_sideways = []
for repetition in range(3):
    start_state = initial_state
    solver_sideways.append(SidewaysMove(obj, max_sideways_moves=10, max_iterations = 100))
    results_sideways.append(solver_sideways[-1].solve(start_state))
    print(f"Results of repetition {repetition+1}:")
    print(results_sideways[repetition]) # Output: Runtime, Total Iterations, Initial Objective, Final Objective, Random Successor Iterations
    print("Total Sideways Moves:", results_sideways[repetition].metadata["sideways_moves"]) # Output: Total Sideways Moves
    
    plt.plot(results_sideways[repetition].objective_history) # Output: Plot of Objective Value over Iterations
    
    draw_state(results_sideways[repetition].final_state) # Output: Visualization of Final State
    print("=" * 50)

plt.title('Objective History')
plt.xlabel('Iterations')
plt.ylabel('Objective')
plt.show()
    
```

```python
# Algorithm 1.4: Random Restart Hill Climbing
solver_ranres = []
results_ranres = []
for repetition in range(3):
    start_state = initial_state
    solver_ranres.append(RandomRestart(obj, max_restarts=5, max_iterations_per_restart=100, initial_state=start_state))
    results_ranres.append(solver_ranres[-1].solve())
    print(f"Results of repetition {repetition+1}:")
    print(results_ranres[repetition]) # Output: Runtime, Total Iterations, Initial Objective, Final Objective, Random Successor Iterations
    print("Final Objective Value:", results_ranres[repetition].metadata["final_objective_value"]) # Output: Final Objective Value
    print("Total Restarts:", results_ranres[repetition].metadata["total_restarts"]) # Output: Total Restarts
    print("Iterations per Restart:", results_ranres[repetition].metadata["iterations_per_restart"]) # Output: Iterations per Restart
    
    plt.plot(results_ranres[repetition].objective_history) # Output: Plot of Objective Value over Iterations
    
    draw_state(results_ranres[repetition].final_state) # Output: Visualization of Final State
    print("=" * 50)

plt.title('Objective History')
plt.xlabel('Iterations')
plt.ylabel('Objective')
plt.show()
    
```

```python
# Algorithm 2: Simmulated Annealing
obj = MaxValueObjective() # Setting up the objective function
schedule = RationalSchedule(100, 0.001) # Setting up the scheduling function

solver_simann = []
results_simann = []
fig_obj, ax_obj = plt.subplots()
fig_prob, ax_prob = plt.subplots()

for repetition in range(3):
    start_state = initial_state
    solver_simann.append(SimulatedAnnealing(obj, schedule, max_iterations=100))
    results_simann.append(solver_simann[-1].solve(start_state))

    print(f"Results of repetition {repetition+1}:")

    ax_obj.plot(results_simann[repetition].objective_history, label=f"Repetition {repetition + 1}")

    ax_prob.plot(results_simann[repetition].metadata['probability_history'], label=f"Repetition {repetition + 1}")

    print(results_simann[repetition])
    draw_state(results_simann[repetition].final_state)

    print("=" * 50)

ax_obj.set_title("Objective History")
ax_obj.set_xlabel("Iterations")
ax_obj.set_ylabel("Objective")
ax_obj.legend()

ax_prob.set_title("Probability History")
ax_prob.set_xlabel("Iterations")
ax_prob.set_ylabel("Probability")
ax_prob.legend()

plt.show()
```
